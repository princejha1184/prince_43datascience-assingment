### NumPy (Q35–Q38)

### Q35

```python
np.random.seed(42)
marks = np.random.randint(20, 101, (12, 4))
averages = marks.mean(axis=1)
test_mean = marks.mean(axis=0)
test_std = marks.std(axis=0)
passed = np.all(marks >= 40, axis=1)
high_average = np.where(averages >= 75)[0] + 1
most_variable_test = np.argmax(test_std) + 1
print("Marks:\n", marks)
print("Student averages:", averages)
print("Test means:", test_mean)
print("Test standard deviations:", test_std)
print("Students passing all tests:", passed.sum())
print("Students with average >= 75:", high_average)
print("Most variable test:", most_variable_test)
```

### Q36

```python
a = np.arange(21, 46).reshape(5, 5)
b = a[1:4, 1:4]
c = b.T
d = b @ c
row_sums = d.sum(axis=1)
maximum = row_sums.max()
positions = np.argwhere(row_sums == maximum)
print("Original matrix:\n", a)
print("Central matrix:\n", b)
print("Transpose:\n", c)
print("Product:\n", d)
print("Row sums:", row_sums)
print("Maximum:", maximum)
print("Positions:", positions)
```

### Q37

```python
quantity = np.array([
    [10, 8, 15, 5],
    [12, 9, 18, 7],
    [20, 10, 12, 8],
    [15, 14, 20, 10],
    [18, 12, 16, 9],
    [22, 11, 25, 12]
])
price = np.array([150, 240, 90, 320])
daily_revenue = quantity @ price
product_quantity = quantity.sum(axis=0)
product_revenue = product_quantity * price
new_price = np.where(product_quantity > 100, price * 0.9, price)
new_revenue = quantity @ new_price
print("Daily revenue:", daily_revenue)
print("Product quantities:", product_quantity)
print("Product revenue:", product_revenue)
print("Highest revenue day:", daily_revenue.argmax() + 1)
print("Best-selling product:", product_quantity.argmax() + 1)
print("New prices:", new_price)
print("New daily revenue:", new_revenue)
print("New total revenue:", new_revenue.sum())
print("Original prices:", price)
```

### Q38

```python
np.random.seed(42)
a = np.random.normal(50, 12, 100)
outside = np.sum((a < 30) | (a > 70))
cleaned = np.where(a < 30, 30, np.where(a > 70, 70, a))
print("Outside values:", outside)
print("Original mean:", a.mean())
print("Cleaned mean:", cleaned.mean())
print("Original variance:", a.var())
print("Cleaned variance:", cleaned.var())
print("Original standard deviation:", a.std())
print("Cleaned standard deviation:", cleaned.std())
print("Original minimum:", a.min())
print("Cleaned minimum:", cleaned.min())
print("Original maximum:", a.max())
print("Cleaned maximum:", cleaned.max())
```

### Pandas Series (Q39–Q40)

### Q39

```python
branch_a = pd.Series({"Q1": 120, "Q2": 145, "Q4": 180})
branch_b = pd.Series({"Q2": 110, "Q3": 135, "Q4": 155})
branch_c = pd.Series({"Q1": 95, "Q3": 125, "Q4": 140})
result = branch_a + branch_b + branch_c
common = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
common_total = branch_a[common] + branch_b[common] + branch_c[common]
zero_result = branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0)
print("Ordinary addition:\n", result)
print("Common quarters:", common.tolist())
print("Common quarter earnings:\n", common_total)
print("Missing treated as zero:\n", zero_result)
```

### Q40

```python
scores = {"R101": 78, "R102": 65, "R104": 92, "R105": 48}
required = ["R105", "R103", "R101", "R106", "R104", "R102"]
s = pd.Series(scores, index=required)
print("Original Series:\n", s)
print("Missing records:\n", s[s.isnull()])
filtered = s[s.notnull() & (s >= 70)]
print("Scores >= 70:\n", filtered)
d = filtered.to_dict()
new_series = pd.Series(d)
print("Dictionary:", d)
print("New Series:\n", new_series)
print("Values match:", filtered.equals(new_series))
```

## Notes

- `axis=0` works down the rows and calculates results for each column.
- `axis=1` works across the columns and calculates results for each row.
- Pandas Series operations align values by index labels.
- `NaN` represents a missing value.
- A NumPy slice is often a view; `.copy()` creates an independent copy.
